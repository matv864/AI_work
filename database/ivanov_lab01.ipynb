### Необязательное Expert homework

Исследуйте, как ведёт себя NOT IN, если в наборе сравнения появляется NULL. \
Сделайте минимальный пример и объясните результат своими словами.

по примеру ниже видим, что проверка на наличи в множестве ведёт себя странно

```python
query = '''
SELECT (1 IS NULL), (1 IS NOT NULL), (1 NOT IN (2, NULL))
;
'''
pd.read_sql_query(query, connection)
```

проверку мы можем выразить в другом виде, то есть \
a NOT IN (b, NULL)   =>    a <> b AND a <> NULL \
и как раз эта часть (a <> NULL) и даёт None, \
потому что NULL - это отсутствие присутствия, а не присутстыие отсутствия (0)
поэтому необходимо использовать проверку на нулы и синтаксис IS NOT NULL

```python
query = '''
SELECT (1 <> NULL), (1 IS NOT NULL), (1 NOT IN (2, NULL))
;
'''
pd.read_sql_query(query, connection)
```
