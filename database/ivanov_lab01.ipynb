задание: Если SQL для вас давно знаком: решите Expert B минимум двумя способами и объясните, какой вариант легче читать и поддерживать. \
ответ: третий вариант проще в чтении и поддержке, потому что написан более лаконичнее и используется лишь одна логическая операция, в остальных вариантах с 3-4 операциями легко запустаться
